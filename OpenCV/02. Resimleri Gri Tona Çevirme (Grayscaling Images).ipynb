# **Resimleri Gri Tona Çevirme**

Bu derste şunları öğreneceğiz:
1. Renkli bir görüntüyü gri tonlamalıya dönüştürme
2. Gri tonlamalı ve renkli görüntüler arasındaki boyutlardaki değişimi görme



```python
import cv2
from matplotlib import pyplot as plt
```

```python
# Define our imshow function 
def imshow(title = "Image", image = None, size=10):  
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

```

```python
# Load our input image
image = cv2.imread('../files/images/aliyukari.jpg')

imshow("Ali Koç Yukarıya Bakıyor", image)
```

```python
image.shape[:2] # (1200, 1920) (height, width)
def imshow(title = "", image = None, size = 10):
 

    h, w = image.shape[0], image.shape[1]
    aspect_ratio = w/h
# size değişkeni 10 dan farklı bir değer verilirse görüntülenen resim boyutları değişecek
# yani daha büyük veya küçük görünecek. Sadece görüntülemede değişiyor. Orijinal büyüklüğü korunuyor.
# örneğin size=4
    size=5
    plt.figure(figsize=(size * aspect_ratio, size))
    #figsize=(size * aspect_ratio, size) şeklin genişliğini ve yüksekliğini inç cinsinden ayarlar.
    #size  şeklin istenen yüksekliğini temsil eden bir değişkendir.
    #aspect_ratio, genişliğin yüksekliğe istenen oranını temsil eden bir değişkendir. 
    #Genişlik, boyutun aspect_ratio ile çarpılmasıyla hesaplanır.
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

imshow("Ali Koç Yukarıya Bakıyor", image)
```

```python
# Gri tonlamaya dönüştürmek için cvtColor kullanıyoruz
# İki değişken alır, ilki girdi görüntüsü
# İkincisi ise renk uzayı dönüştürme formatıdır
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

imshow("Converted to Grayscale", gray_image)
```
