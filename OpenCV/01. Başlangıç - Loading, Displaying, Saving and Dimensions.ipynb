### **Resimleri Yükleme(Loading Images)**

```python
# 'imread' kullanarak bir görüntü yükleyin, görüntünün yolunu belirtin
image = cv2.imread('../files/images/castara.jpeg')
print("dtype:", image.dtype)
print("shape:", image.shape)
# image bir NumPy ndarray ve görüntü verisini içeren çok-boyutlu bir dizi olarak kullanılır.
```

### **Resimleri Görüntüleme (Displaying Images)**

```python
from matplotlib import pyplot as plt
#Resmi matplotlib ile göster
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
#plt.show()

#Jupyter Notebook/JupyterLab içinde Çoğu durumda plt.show() çağrısına gerek kalmaz. Son hücrede oluşturulan figürler otomatik olarak render edilir 
#ve çıktıda görüntülenir.
#Standalone Python scripti plt.show() olmadan program çalışır ve grafik penceresi açılmaz; script hemen sonlanır ve pencere hiç oluşmayabilir.
```

Resimlerimizi görüntülemeyi daha basit ve kolay hale getirmek için basit bir fonksiyon oluşturalım.

```python
def imshow(title = "", image = None):
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

```python
#Deneyelim
imshow("Displaying Our First Image", image)
```

### **Resimleri Kaydetme (Saving Images)**

```python

# Simply use 'imwrite' specificing the file name and the image to be saved
cv2.imwrite('output.jpg', image)

#hangi dizinde (klasörde) çalışıyorsanız o dizine kayıt yapacaktır. Çalıştığınız dizini görmek için şağıdaki kodu kullanabilirsiniz
#import os
#cwd = os.getcwd()
#print(cwd)
```

```python
# Or save it as a PNG (Portable Network Graphics) which is a Lossless bitmap image format
cv2.imwrite('output.png', image)
```
