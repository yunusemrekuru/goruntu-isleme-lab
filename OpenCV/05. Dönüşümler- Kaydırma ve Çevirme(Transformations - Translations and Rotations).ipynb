````python
import cv2
import numpy as np
from matplotlib import pyplot as plt


def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()

````

### **Translations**

This an affine transform that simply shifts the position of an image. (left or right).

We use cv2.warpAffine to implement these transformations.
### **Dönüşümler**

Bu, bir görüntünün konumunu basitçe kaydıran bir affine dönüşümüdür. (sola veya sağa).

Bu dönüşümleri uygulamak için cv2.warpAffine kullanırız.


```cv2.warpAffine(image, T, (width, height))```

![](https://raw.githubusercontent.com/rajeevratan84/ModernComputerVision/main/warp.png)

**Kullanım:**\
dst = cv2.warpAffine(src, M, dsize[, dst[, flags[, borderMode[, borderValue]]]])\

**Parametreler:**
* src → Giriş görüntüsü (numpy array).
* M → 2x3 dönüşüm matrisi (ör: cv2.getRotationMatrix2D veya cv2.getAffineTransform ile elde edilir).
* dsize → Çıktı görüntüsünün boyutu (width, height) formatında.
* flags (opsiyonel) → Enterpolasyon yöntemi:
    * cv2.INTER_NEAREST → en yakın komşu (hızlı, düşük kalite)
    * cv2.INTER_LINEAR → lineer interpolasyon (varsayılan)
    * cv2.INTER_CUBIC → bicubic interpolasyon (daha kaliteli, daha yavaş)
* borderMode (opsiyonel) → Görüntü sınırlarının dışında kalan pikseller:
    * cv2.BORDER_CONSTANT (varsayılan, dışı tek renk)
    * cv2.BORDER_REFLECT, cv2.BORDER_WRAP gibi diğer modlar
    * borderValue → BORDER_CONSTANT seçildiğinde kenarların rengi (varsayılan siyah (0,0,0)).

````python
# Resmi yükle
image = cv2.imread('../files/images/america2.jpg')
imshow("Original", image)

# Resmin yükseklik ve genişliğini kaydet
height, width = image.shape[:2]

# yükseklik ve genişliğin 4'te 1'i kadar kaydırıyoruz
quarter_height, quarter_width = height/4, width/4

# Dönüşüm
#       | 1 0 Tx |
#  T  = | 0 1 Ty |
# T dönüşüm matrisidir
# Örneğin
# 50px sağa, 30px aşağı kaydırma matrisi
#M = np.float32([[1, 0, 50], 
#                [0, 1, 30]])

T = np.float32([[1, 0, quarter_width], [0, 1,quarter_height]])

# T matrisini kullanarak görüntüyü dönüştürmek için warpAffine kullanırız
img_translation = cv2.warpAffine(image, T, (width, height))
imshow("Translated", img_translation)
````

### **Rotasyonlar**

```cv2.getRotationMatrix2D(rotation_center_x, rotation_center_y, angle of rotation, scale)```

cv2.getRotationMatrix2D(merkez, açı, ölçek katsayısı)
ölçek katsayısı (1.0 → aynı boyut, 0.5 → küçültme, 2.0 → büyütme).



````python
# Resmi yükleyelim
image = cv2.imread('../files/images/america2.jpg')
height, width = image.shape[:2]

# Görüntüyü merkezi etrafında döndürmek için merkez noktasını kullanalım
rotation_matrix = cv2.getRotationMatrix2D((width/2, height/2), 90, 1)

# Resmimizi, döndürme matrisini ve istediğimiz nihai genişlik ve yüksekliği girin
rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))
imshow("Rotated 90 degrees with scale = 1", rotated_image)
````

````python
# Resmi merkezi etrafında döndürmek için merkezi yarısı olarak belirleyin
rotation_matrix = cv2.getRotationMatrix2D((width/2, height/2), 90, 0.5)
print(rotation_matrix)
# Resmimizi, döndürme matrisini ve istediğimiz nihai genişlik ve yüksekliği girin
rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))
imshow("Rotated 90 degrees with scale = 0.5", rotated_image)
````

### **Cv2.transpose** ile rotasyonlar (daha az esnek)
OpenCV’de bir görüntünün veya matrisin transpozunu almak için kullanılır.\
Yani satır ve sütunları yer değiştirir.
Transpoz bir görüntüyü saat yönünde 90° döndürmekle aynı değildir direkt olarak.



````python
rotated_image = cv2.transpose(image)
imshow("Original", image)
imshow("Rotated using Transpose", rotated_image)
````

````python
rotated_image = cv2.transpose(image)
rotated_image = cv2.transpose(rotated_image)

imshow("Rotated using Transpose", rotated_image)
````
