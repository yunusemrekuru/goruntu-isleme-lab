### **Bulanıklığı Tespit Etmek için Laplacian çekirdeği ile Konvolüsyon yaparız.**

Bir görüntünün gri tonlamasını alıp Laplacian kernel (3 x 3 kernel) ile Konvolüsyon yaparız:

Bulanıklığı ölçmek için yanıt çıktısının varyansını alırız.

Laplacian bir görüntünün 2. türevidir ve bu nedenle bir görüntünün hızlı yoğunluk değişiklikleri içeren alanlarını vurgular. Bu nedenle Kenar Algılamada kullanılır. 

Bu nedenle düşük varyans, görüntüde çok az kenar olduğu anlamına gelebilir, bu da bulanık olabileceği anlamına gelir, çünkü ne kadar çok bulanıklık varsa o kadar az kenar vardır.

```python
import cv2
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

#### **Bulanık Görüntüler Üretelim**

```python
# Girdi resmimizi yükleyin
image = cv2.imread('../files/images/america37.jpg')
imshow("Original Image", image)


# Gaussian Blur filtresi uygulanıyor. (5,5) -> filtre kernel boyutunu belirtiyor. Küçük olduğu için resim az bulanık olur.
blur_1 = cv2.GaussianBlur(image, (5,5), 0)
imshow('Blurred Image 1', blur_1) 

# Daha büyük bir kernel → daha fazla bulanıklık.
blur_2 = cv2.GaussianBlur(image, (9,9), 0)
imshow('Blurred Image 2', blur_2) 

# Daha da büyük bir kernel → daha da fazla bulanıklık.
blur_3 = cv2.GaussianBlur(image, (25,25), 0)
imshow('Blurred Image 3', blur_3) 
```
