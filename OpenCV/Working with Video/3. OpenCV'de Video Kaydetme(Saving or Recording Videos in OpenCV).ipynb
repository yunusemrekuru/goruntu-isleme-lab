# **OpenCV'de Video Kaydı**

**Bu derste şunları öğreneceğiz:**
1. cv2.VideoWriter() kullanarak bir video kaydedin

out = cv2.VideoWriter('drummer.avi', 
                      cv2.VideoWriter_fourcc('M','J','P','G'), 
                      30, 
                      (w, h), 
                      False)
**VideoWriter** nesnesini kullanırken. Belirtmemiz gerekenler:
- Kaydedilecek dosyanın adı ve uzantısı. Burada .avi formatında bir video dosyası kaydedilecek.
- cv2.VideoWriter_fourcc('M','J','P','G')
    - VideoWriter_fourcc fonksiyonu, bu dört karakteri sayıya çevirir ve VideoWriter’a aktarır.
    - Kullanılacak codec bilgisidir.'MJPG' (Motion JPEG), her kareyi JPEG resmi gibi sıkıştırarak kaydeder.
    - 30 FPS (frame per second). Yani video saniyede 30 kareden oluşacak.
 - (w, h)
    - Çıkış videosunun çözünürlüğü.
    - w = genişlik, h = yükseklik.
    - Bu değerler kaydedilecek karelerle birebir aynı olmalı, yoksa hata oluşur.
 - False
   - Renkli mi, gri tonlamalı mı kaydedileceğini belirtir.
   - True → renkli (BGR formatında)
   - False → siyah-beyaz (gri tonlamalı)

```python
import cv2

# Load video clip
cap = cv2.VideoCapture('../../files/videos/drummer.mp4')

# Get the height and width of the frame (required to be an interger)
w = int(cap.get(3)) 
h = int(cap.get(4))

# Define the codec and create VideoWriter object. The output is stored in '*.avi' file.
out = cv2.VideoWriter('drummer.avi', cv2.VideoWriter_fourcc('M','J','P','G'), 30, (w, h), False)

while(1):

    ret, frame = cap.read()

    if ret == True:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        out.write(gray)
    else:
        break
        print("Converted Video to Grayscale..")
        
cap.release()
out.release()
```
