# **OpenCV'de Video Dosyalarını Açmak **

**Bu derste şunları öğreneceğiz:**
1. cv2.VideoCapture() kullanarak bir video açın

**NOT:**

AVI video formatları OpenCV'nin tüm sürümleri tarafından evrensel olarak desteklenir, bu nedenle çapraz platform uyumluluğuna öncelik veriyorsanız .AVI dosyalarını kullanın.

**Yaygın olarak desteklenen diğer Video Codec'leri**
1. MPEG4 veya MP4
2. FLV

Bir sistemde kullanabileceğiniz codec bileşenlerinin tam listesi, yalnızca hangisini yüklediğinize bağlıdır.

````python
import cv2

cap = cv2.VideoCapture('../../files/videos/drummer.mp4')

while cap.isOpened():
    
    ret, frame = cap.read()
    
    # if frame is read correctly ret is True
    if not ret:
        print("Stream stopped.")
        break

    cv2.imshow('Video', frame)
    
    if cv2.waitKey(1) == ord('q'):
        break
    
        
cap.release()
cv2.destroyAllWindows()
````

### Check Video Streams before processing

````python
import cv2

cap = cv2.VideoCapture('../../files/videos/drummer.mp4')

# Check Video Streams before processing
if not cap.isOpened():
    print("Cannot open video")
    exit()

while cap.isOpened():
    
    ret, frame = cap.read()
    
    # çerçeve doğru okunmuşsa ret True'dur
    if not ret:
        print("Stream stopped.")
        break

    cv2.imshow('Video', frame)
    
    if cv2.waitKey(1) == ord('q'):
        break
    
        
cap.release()
cv2.destroyAllWindows()
````

## ```cap.get(id)``` 
Is a method where id is a number from 0 to 18. Each number denotes a property of the video (if it is applicable to that video)

````python
import cv2

cap = cv2.VideoCapture('../../files/videos/drummer.mp4')

for i in range(0,18):
    print(cap.get(i))
````
