# **Web Kameranızı OpenCV'de Kullanma**

**Bu derste şunları öğreneceğiz:**
1. cv2.VideoCapture() kullanarak web kameranızdan bir video akışı açın
2. Web kamerası girişinize bir taslak işlevi uygulayın

## **Web Kameranızdan Video Açın**

```python
import cv2
import numpy as np

# Web kamerasını başlatın, cap VideoCapture tarafından sağlanan nesnedir
cap = cv2.VideoCapture(0)
#cv2.VideoCapture(0) satırındaki 0, bilgisayara bağlı olan kamera aygıtını belirtir.
#0: Varsayılan (genellikle dizüstü bilgisayarın dahili) kamerayı açar.
#1: Bilgisayara bağlı ilk harici kamerayı açar.
#2: Birden fazla kamera varsa ikinci harici kamerayı açar, ve bu şekilde devam eder.
i=1
while True:
    # Başarılı olup olmadığını gösteren bir boolean içerir (ret)
    # Ayrıca web kamerasından toplanan görüntüleri de içerir (çerçeve)
    ret, frame = cap.read()
    
    cv2.imshow('Our Webcam Video', frame)
    print("frame ",i)
    i+=1
    if cv2.waitKey(1) == 13: #13 is the Enter Key
        break
        
# Release camera and close windows
cap.release()
cv2.destroyAllWindows()
```

## **Kendinizin Canlı Bir Eskizini Oluşturmak İçin Web Kameranızın Akışını Kullanın**

```python
import cv2
import numpy as np

# Eskiz oluşturma 
def sketch(image):
    # Görüntüyü gri tonlamaya dönüştür
    img_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Guassian Blur kullanarak görüntüyü temizleyin
    img_gray_blur = cv2.GaussianBlur(img_gray, (5,5), 0)
    
    # Kenarları ayıkla
    canny_edges = cv2.Canny(img_gray_blur, 10, 70)
    
    # Görüntüyü ters çevirerek ikili hale getirin 
    ret, mask = cv2.threshold(canny_edges, 70, 255, cv2.THRESH_BINARY_INV)
    return mask


# Web kamerasını başlatın, cap VideoCapture tarafından sağlanan nesnedir
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()
    # Çerçevemizi doğrudan cv2.imshow() içinde sketch fonksiyonumuza aktarıyoruz
    cv2.imshow('Our Live Sketcher', sketch(frame))
    if cv2.waitKey(1) == 13: #13 is the Enter Key
        break
        
# Kamera değişkenini serbest bırakın ve pencereleri kapatın
cap.release()
cv2.destroyAllWindows()      
```
