### **RTSP Gerçek Zamanlı Akış Protokolü**

RSPT, IP ağları üzerinden akışlı multimedyanın verimli bir şekilde teslim edilmesi ihtiyaçlarını karşılamak için tasarlanmış istemci-sunucu multimedya sunum kontrol protokolüdür. RTSP için kullanılan temel protokol RTP protokolüdür. 

RTSP 1996 yılında RealNetworks, Netscape ve Columbia Üniversitesi tarafından geliştirilmiştir. İstemci ve sunucu arasında gerçek zamanlı multimedya verilerini (örneğin ses/video) aktarmak için kullanılan bir protokoldür. Genellikle bir İstemci talepte bulunur ve sunucu bu protokol üzerinden veri ile talebe yanıt verir. Bir akış protokolü olduğundan, gerçek zamanlı özelliğe sahiptir, bu da verilerin aynı anda gerçek zamanlı olarak aktarıldığı ve işlendiği anlamına gelir. Burada multimedya verileri Gerçek Zamanlı Aktarım Protokolü (RTP) paketleri altında kapsüllenir. Yani işi yapan RTSP değil, RTP'dir. Çoğu

```python
import cv2 

# Ücretsiz Test RSTP Bağlantımız
# CCTV IPTV Kameralarınızı bir RSTP Akışı çıkışı verecek şekilde yapılandırabilirsiniz
cap = cv2.VideoCapture("rtsp://wowzaec2demo.streamlock.net/vod/mp4:BigBuckBunny_115k.mov")

while(1):
    ret, frame = cap.read()

    cv2.imshow('RTSP Stream', frame)
    
    if cv2.waitKey(1) == 13: #13 Enter Tuşudur
        break
        
# Release camera and close windows
cap.release()
cv2.destroyAllWindows()      
```
