## **Tek Ekran Görüntüsü Yakala**

```python
#Tüm ekranın görüntüsünü alır (fullscreen).
#Aldığı görüntüyü fullscreen.png dosyasına kaydeder.
import pyscreenshot as ImageGrab

# tam ekran yakala
im = ImageGrab.grab()

# görüntü dosyasını kaydet
im.save('fullscreen.png')
```

## **Ekrandan Video Yakala**

```python
# Belirli bir ekran bölgesini (bbox=(0,0,300,300)) sürekli yakalar.
import numpy as np
from PIL import ImageGrab
import cv2
import time

last_time = time.time()

while(True):
    
    frame = np.array(ImageGrab.grab(bbox=(0,0,300,300)))
    
    # Tam ekran olması için
    #frame = np.array(ImageGrab.grab())
    
    # FPS oranını görüntüleme. Her döngüde geçen süreyi ölçüp FPS (frames per second) hesaplar ve ekrana yazdırır.
    FPS = 1.0 / (time.time()-last_time)
    print('FPS = {}'.format(FPS))
    last_time = time.time()
    
    # Ekran Görüntüsü Her kareyi numpy array’e çevirip OpenCV (cv2.imshow) ile gösterir.
    cv2.imshow('window', cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
    
    if cv2.waitKey(1) == 13: #13 is the Enter Key
        print("Exited...")
        break
        
cv2.destroyAllWindows()
```

```python
# mss, daha hızlı çalışan bir ekran yakalama kütüphanesidir.
# burada ekranın belirli bir bölgesi (top=100, left=0, width=400, height=300) sürekli kaydedilir.
# Görüntü cv2.imshow ile ekrana verilir.
# her 30 karede bir FPS hesaplanır ve yazdırılır.
import numpy as np
import cv2
from mss import mss
from PIL import Image
import time

bounding_box = {'top': 100, 'left': 0, 'width': 400, 'height': 300}

sct = mss()
frame_count = 0 
last_time = time.time()

while True:
    frame_count += 1
    sct_img = sct.grab(bounding_box)
    cv2.imshow('screen', np.array(sct_img))

    # Display FPS rate
    if frame_count % 30 == 0:
        FPS = 1.0 / (time.time()-last_time)
        print('FPS = {}'.format(FPS))
    last_time = time.time()    

    if cv2.waitKey(1) == 13: #13 is the Enter Key
        print("Exited...")
        break
        
cv2.destroyAllWindows()
```
