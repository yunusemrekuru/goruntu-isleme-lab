# **Aritmetik ve Bitsel İşlemler**

#### **Bu derste şunları öğreneceğiz:**
1. Aritmetik İşlemler
2. Bitsel İşlemler

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Define our imshow function 
def imshow(title = "Image", image = None, size = 10):
    w, h = image.shape[0], image.shape[1]
    aspect_ratio = w/h
    plt.figure(figsize=(size * aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

## **Aritmetik İşlemler**

Bunlar, renk yoğunluğuna doğrudan ekleme veya çıkarma yapmamızı sağlayan basit işlemlerdir.

İki dizinin eleman başına işlemini hesaplar. Genel etki parlaklığı artırmak veya azaltmaktır.

```python
# cv2.imread dosyasına virgül sıfır eklendiğinde görüntümüz gri tonlamalı bir görüntü olarak yüklenir
image = cv2.imread('../files/images/america5.jpg', 0)
imshow("Grayscaled Image",image)
print(image)

# Birlerden oluşan bir matris oluşturun, ardından bunu 100'lük bir ölçekleyici ile çarpın
# Bu, tüm değerleri 100 olan görüntümüzün aynı dimesiyonlarını içeren bir matris verir
M = np.ones(image.shape, dtype = "uint8") * 100 

```

```python
print(M)
```

#### **Parlaklık Arttırma**

```python
# Bunu, M matrisini görüntümüze eklemek için kullanırız
# Parlaklıktaki artışa dikkat edin
added = cv2.add(image, M)
imshow("Increasing Brightness", added)

# Şimdi sadece eklersek sonuç;
added2 = image + M 
imshow("Simple Numpy Adding Results in Clipping", added2)
```

#### **Parlaklık Azaltma**

```python
# Aynı şekilde çıkarabiliriz de
# Parlaklıktaki azalmaya dikkat edin
subtracted = cv2.subtract(image, M)
imshow("Subtracted", subtracted)

subtracted = image - M 
imshow("Subtracted 2", subtracted)
```

## **Bitsel İşlemler ve Maskeleme**

Bu işlemleri göstermek için bazı basit görüntüler oluşturalım

```python
# Kare yapmak
square = np.zeros((800, 800), np.uint8)
cv2.rectangle(square, (50, 50), (250, 250), 255, -2)
imshow("square", square)

# Bir elips yapmak
ellipse = np.zeros((300, 300), np.uint8)
cv2.ellipse(ellipse, (150, 150), (150, 150), 30, 0, 180, 255, -1)
imshow("ellipse", ellipse)
```

### ** AND, OR, XOR ve NOT gibi bazı bitsel işlemlerle denemeler yapma**

```python
# Yalnızca kesiştikleri yerleri gösterir
And = cv2.bitwise_and(square, ellipse)
imshow("AND", And)

# Kare veya elipsin nerede olduğunu gösterir
bitwiseOr = cv2.bitwise_or(square, ellipse)
imshow("bitwiseOr", bitwiseOr)

# İkisinden birinin tek başına bulunduğu yerleri gösterir
bitwiseXor = cv2.bitwise_xor(square, ellipse)
imshow("bitwiseXor", bitwiseXor)

# Karenin parçası olmayan her şeyi gösterir
bitwiseNot_sq = cv2.bitwise_not(square)
imshow("bitwiseNot_sq", bitwiseNot_sq)

# Son işlemin görüntüyü tamamen ters çevirdiğine dikkat edin
```
